```python
# Create a dictionary using the given keys, with the same initial value for every key.
# This is useful when you want to quickly build a dictionary with predefined keys.
user = dict.fromkeys(["ID", "Name", "Age", "City"], "Nill")
print(user)
```

```python
# Create the same dictionary structure, but use None to represent an empty/unknown value.
# None is different from the string 'Nill': it represents the absence of a value.
user2 = dict.fromkeys(["ID", "Name", "Age", "City"], None)
print(user2)
```

```python
# Practice converting string values while building a new dictionary with a normal for loop.
# Only values that are strings are processed.
# This cell is an earlier practice attempt; the later cells show the corrected approach.
user3 = {"ID": 1, "Name": "John", "Age": 30, "City": "Berlin"}
new_dict = {}
for k, v in user3.items():
    if isinstance(v, str):
        new_dict[k.upper] = v.capitalize
        
```

```python
# Dictionary comprehension: create a new dictionary from only the string values in user3.
# k.upper() converts each selected key to uppercase.
# v.capitalize() changes the first character of each selected string value to uppercase.
user3 = {"ID": 1, "Name": "John", "Age": 30, "City": "Berlin"}
user_str = {
    k.upper(): v.capitalize()
    for k, v in user3.items()
    if isinstance(v, str)
}

print(user_str)
```

```python
# A more realistic user record containing both string and non-string values.
# Build a cleaned dictionary containing only string fields.
# Each selected key is converted to uppercase and each string value is capitalized.
user4 = {
    "id": 101,
    "name": "john doe",
    "age": 28,
    "city": "lagos",
    "email": "JOHN@EXAMPLE.COM",
    "is_active": True,
    "country": "nigeria"
}

clean_user = {}
for i,j in user4.items():
    if isinstance(j, str):
        clean_user[i.upper()]=j.capitalize()

print(clean_user)
```

```python
# Synthetic coastal monitoring data for one station.
# The dataset combines several Python data structures:
# - tuple: fixed station location
# - nested dictionaries: measurements organized by parameter and month
# - dictionary: monthly shoreline-change values
# - sets inside a dictionary: hazards recorded for each month
# These values are for Python practice and are not real environmental observations.
station = {
    "station_id": "COAST001",

    "location": ("Lagos", "Nigeria"),

    "measurements": {
        "salinity": {
            "January": 34.5,
            "February": 35.1,
            "March": 34.8,
            "April": 35.4
        },

        "temperature": {
            "January": 28.4,
            "February": 29.1,
            "March": 28.7,
            "April": 29.3
        },

        "pH": {
            "January": 7.9,
            "February": 8.1,
            "March": 8.0,
            "April": 7.8
        }
    },

    "shoreline_change": {
        "January": -0.4,
        "February": -0.6,
        "March": -0.5,
        "April": -0.7
    },

    "hazards": {
        "January": {"flooding"},
        "February": set(),
        "March": {"pollution"},
        "April": {"flooding"}
    }
}
```

```python
# Extract the monthly salinity dictionary from the nested station data.
# The result keeps the month connected to its salinity value.
salinity_value = station["measurements"]["salinity"]
print(salinity_value)
```

```python
# Start an accumulator at zero so the salinity values can be added one at a time.
# Loop through each month and add its salinity measurement to the running total.
total = 0
for month, value in salinity_value.items():
    total += value
print(total)
```

```python
# Divide the total salinity by the number of monthly measurements to calculate the average.
# round(..., 2) keeps the result to two decimal places.
avrg_salinity = round((total/len(salinity_value)), 2)
print(avrg_salinity)
```

```python
# Keep track of the highest salinity value found so far and the month where it occurred.
# Start at 0 so the first positive salinity measurement becomes the initial highest value.
# count records how many times a new highest value is found during the loop.
highest_value = 0
highest_month = ""
count = 0
for month, value in salinity_value.items():
    if highest_value < value:
        highest_value = value
        highest_month = month
        count += 1

print(highest_value)
print(highest_month)
```

```python
# Keep track of the lowest salinity value found so far and its month.
# Start with a large value so the first salinity measurement can replace it.
# count records how many times a new lowest value is found during the loop.
lowest_value = 999
lowest_month = ""
count = 0
for month, value in salinity_value.items():
    if lowest_value > value:
        lowest_value = value
        lowest_month = month
        count += 1

print(lowest_value)
print(lowest_month)
```

```python
# Extract the monthly temperature dictionary from the station's measurements.
temperature = station["measurements"]["temperature"]
print(temperature)
```

```python
# Accumulate all monthly temperature values into one total.
# Divide by the number of measurements to calculate the average temperature.
# Round the result to two decimal places.
total_temp_value = 0
for temp_month, temp_value in temperature.items():
    total_temp_value += temp_value
avrg_total_temp_value = round((total_temp_value/len(temperature)), 2)
print(avrg_total_temp_value)
```

```python
# Find the highest monthly temperature and remember the month associated with it.
# The month is stored separately so the measurement remains linked to its time.
highest_temp = 0
highest_temp_month = ""
for temp_month, temp_value in temperature.items():
    if highest_temp < temp_value:
        highest_temp = temp_value
        highest_temp_month = temp_month

print(highest_temp)
print(highest_temp_month)
```

```python
# Find the lowest monthly temperature and remember the month associated with it.
# Start with a large value so the first measurement becomes the initial lowest value.
lowest_temp = 999
lowest_temp_month = ""
for temp_month, temp_value in temperature.items():
    if lowest_temp > temp_value:
        lowest_temp = temp_value
        lowest_temp_month = temp_month

print(lowest_temp)
print(lowest_temp_month)
```

```python
# NumPy is used here to demonstrate numerical array operations.
# Convert the monthly temperature and salinity values into lists first.
# Then convert the lists into NumPy arrays so subtraction can happen element by element.
# Note: converting to arrays this way removes the month labels from the result.
import numpy as np

temperature = station["measurements"]["temperature"]
salinity = station["measurements"]["salinity"]

temperature_value = []
salinity_value = []
for temp_month, temp_value in temperature.items():
    temperature_value.append(temp_value)
for sal_month, sal_value in salinity.items():
    salinity_value.append(sal_value)

temp_arr = np.array(temperature_value)
sal_arr = np.array(salinity_value)

print(temp_arr)
print(sal_arr)
difference = sal_arr - temp_arr
print(difference)
```

```python
# Compare salinity and temperature while preserving the month as the dictionary key.
# The common month key is used to retrieve the matching temperature measurement.
# Store each calculated difference under its corresponding month.
difference = {}
temperature = station["measurements"]["temperature"]
salinity = station["measurements"]["salinity"]

for month, sal_value in salinity.items():
    difference[month] = sal_value - temperature[month]

print(difference)
```

```python
# Re-create the synthetic station dataset so the hazard-analysis section has the station data available.
# The hazards are represented as sets because a month can contain one or more distinct hazard types.
station = {
    "station_id": "COAST001",

    "location": ("Lagos", "Nigeria"),

    "measurements": {
        "salinity": {
            "January": 34.5,
            "February": 35.1,
            "March": 34.8,
            "April": 35.4
        },

        "temperature": {
            "January": 28.4,
            "February": 29.1,
            "March": 28.7,
            "April": 29.3
        },

        "pH": {
            "January": 7.9,
            "February": 8.1,
            "March": 8.0,
            "April": 7.8
        }
    },

    "shoreline_change": {
        "January": -0.4,
        "February": -0.6,
        "March": -0.5,
        "April": -0.7
    },

    "hazards": {
        "January": {"flooding"},
        "February": set(),
        "March": {"pollution"},
        "April": {"flooding"}
    }
}
```

```python
# Extract the monthly hazard information from the station data.
# Each month maps to a set containing the hazards recorded during that month.
hazard = station["hazards"]
print(hazard)
```

```python
# Start with an empty set for all unique hazard types.
# update() adds every item from each month's set.
# Because this is a set, repeated hazards such as flooding are stored only once.
all_hazards = set()
for month, hazard_list in hazard.items():
    all_hazards.update(hazard_list)

print(all_hazards)
```

```python
# Visit each month and then each individual hazard recorded in that month.
# The inner loop allows the program to work with each hazard separately.
for month, hazard_list in hazard.items():
    for i in hazard_list:
        print(f"{i} occured in {month}")
```

```python
# Build a frequency dictionary showing how many times each hazard occurred.
# If a hazard has not been seen before, start its count at 1.
# If it already exists, increase its existing count by 1.
hazard_count = {}
count = 0
for month, hazard_list in hazard.items():
    for i in hazard_list:
        if i not in hazard_count:
            hazard_count[i] = 1
        else:
            hazard_count[i] += 1

print(hazard_count)
```
